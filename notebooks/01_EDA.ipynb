# Customer Churn Analytics - Exploratory Data Analysis

## Objective

The purpose of this notebook is to explore the cleaned customer churn dataset,
understand customer behavior, identify patterns associated with churn, and
visualize important factors that may influence customer churn.

The cleaned dataset contains 7,043 customer records and 26 features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/cleaned_customer_churn.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 1. Dataset Overview

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())
```

df.info()

## 2. Data Quality Check

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 3. Churn Distribution

The `Churn` column is the target variable. It indicates whether a customer
left the company.

```python
churn_counts = df["Churn"].value_counts()

print(churn_counts)

churn_counts.plot(kind="bar")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.show()
```

```python
churn_percentage = df["Churn"].value_counts(normalize=True) * 100

print(churn_percentage.round(2))
```

## 4. Churn by Contract Type

This analysis examines whether customer churn differs across contract types.

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print(contract_churn.round(2))
```

```python
contract_churn.plot(kind="bar", stacked=True)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.show()
```

## 5. Churn by Internet Service

This section examines the relationship between internet service type and customer churn.

```python
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

print(internet_churn.round(2))
```

```python
internet_churn.plot(kind="bar", stacked=True)

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Percentage")
plt.show()
```

```python
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

print(payment_churn.round(2))
```

```python
payment_churn.plot(kind="bar", stacked=True)

plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Percentage")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 7. Customer Tenure Analysis

Customer tenure represents the number of months a customer has stayed with the company.

```python
print(df.groupby("Churn")["tenure"].describe())
```

```python
df.boxplot(column="tenure", by="Churn")

plt.title("Customer Tenure by Churn Status")
plt.suptitle("")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.show()
```

## 8. Monthly Charges Analysis

This section compares monthly charges between customers who churned and customers who stayed.

```python
print(df.groupby("Churn")["MonthlyCharges"].describe())
```

```python
df.boxplot(column="MonthlyCharges", by="Churn")

plt.title("Monthly Charges by Churn Status")
plt.suptitle("")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()
```

# 10. Key Findings

The exploratory analysis provides several useful observations about customer churn:

- Churn behavior differs across contract types.
- Customer tenure is an important factor when examining churn.
- Monthly charges show differences between churned and retained customers.
- Internet service type and payment method are useful variables for understanding churn.
- The engineered features provide additional information about customer spending,
  tenure, and service usage.
- These patterns support the use of machine learning for predicting customers
